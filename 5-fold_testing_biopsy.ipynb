```python
from ultralytics import YOLO
import pandas as pd
from pathlib import Path

ROOT = Path(
    "/home/zaman/Code/Breast_Cancer_Detection_and_Monitoring/Cancer.v1i.yolov8/5-fold"
)

results = []

for fold in range(1, 6):

    print(f"\n========== FOLD {fold} ==========\n")

    yaml_file = ROOT / f"fold{fold}.yaml"

    # -------------------------------------------------
    # CHANGE THIS TO YOUR MODEL
    # -------------------------------------------------

    model = YOLO("yolov8n.pt")

    model.train(
        data=str(yaml_file),
        epochs=100,
        imgsz=512,
        batch=4,
        name=f"fold{fold}",
        project="5fold_runs"
    )

    csv_path = (
        Path("5fold_runs")
        / f"fold{fold}"
        / "results.csv"
    )

    df = pd.read_csv(csv_path)

    best_idx = df["metrics/mAP50-95(B)"].idxmax()

    best = df.loc[best_idx]

    precision = best["metrics/precision(B)"]
    recall = best["metrics/recall(B)"]

    f1 = (
        2 * precision * recall
        / (precision + recall + 1e-9)
    )

    results.append({
        "Fold": fold,
        "Epoch": int(best["epoch"]),
        "Precision": precision,
        "Recall": recall,
        "F1": f1,
        "mAP50": best["metrics/mAP50(B)"],
        "mAP50_95": best["metrics/mAP50-95(B)"]
    })

summary = pd.DataFrame(results)

summary.to_csv(
    "5fold_results.csv",
    index=False
)

print("\nSaved 5fold_results.csv")
```

```python
from ultralytics import YOLO
import pandas as pd
from pathlib import Path
```

```python
FOLD = 1
```

```python
ROOT = Path(
    "/home/zaman/Code/Breast_Cancer_Detection_and_Monitoring/Cancer.v1i.yolov8/5-fold"
)

yaml_file = ROOT / f"fold{FOLD}.yaml"

model = YOLO("yolov8n.pt")

model.train(
    data=str(yaml_file),
    epochs=100,
    imgsz=512,
    batch=4,
    project="5fold_runs",
    name=f"fold{FOLD}",
    exist_ok=True
)
```

```python
import pandas as pd
from pathlib import Path

csv_path = Path(f"5fold_runs/fold{FOLD}/results.csv")

df = pd.read_csv(csv_path)

df.columns = df.columns.str.strip()

best_idx = df["metrics/mAP50-95(B)"].idxmax()

best = df.loc[best_idx]

precision = best["metrics/precision(B)"]
recall = best["metrics/recall(B)"]

f1 = 2 * precision * recall / (precision + recall + 1e-9)

result = pd.DataFrame([{
    "Fold": FOLD,
    "Best Epoch": int(best["epoch"]),
    "Precision": precision,
    "Recall": recall,
    "F1": f1,
    "mAP50": best["metrics/mAP50(B)"],
    "mAP50-95": best["metrics/mAP50-95(B)"]
}])

result
```

```python
result.to_csv(
    f"fold_{FOLD}_summary.csv",
    index=False
)

print(f"Saved fold_{FOLD}_summary.csv")
```

```python
#Run only after all the folds are completed
import pandas as pd

files = [
    "fold_1_summary.csv",
    "fold_2_summary.csv",
    "fold_3_summary.csv",
    "fold_4_summary.csv",
    "fold_5_summary.csv"
]

dfs = [pd.read_csv(f) for f in files]

all_folds = pd.concat(dfs, ignore_index=True)

all_folds
```

```python
metrics = [
    "Precision",
    "Recall",
    "F1",
    "mAP50",
    "mAP50-95"
]

summary = []

for metric in metrics:

    summary.append({
        "Metric": metric,
        "Mean": all_folds[metric].mean(),
        "Std": all_folds[metric].std(),
        "Mean ± Std":
            f"{all_folds[metric].mean():.4f} ± "
            f"{all_folds[metric].std():.4f}"
    })

summary_df = pd.DataFrame(summary)

summary_df
```

```python
paper_table = summary_df[["Metric", "Mean ± Std"]]

paper_table
```

```python
import matplotlib.pyplot as plt

metrics = [
    "Precision",
    "Recall",
    "F1",
    "mAP50",
    "mAP50-95"
]

ax = all_folds.set_index("Fold")[metrics].plot(
    kind="bar",
    figsize=(12,6)
)

plt.title("5-Fold Cross Validation Results")
plt.ylabel("Score")
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()

plt.savefig(
    "5fold_comparison.png",
    dpi=300
)

plt.show()
```
